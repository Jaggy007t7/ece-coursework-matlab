```python
import matplotlib.pyplot as plt 
import numpy as np 

t = np.linspace(0,1,1000)

```

1.1 Generate and plot sinusoidal signal of 5V amplitude, 5Hz frequency and zero phase.

```python

f = 5
plt.plot(t,5*np.sin(2*np.pi*f*t))
plt.xlabel("time(t)")
plt.ylabel("Amplitude")
plt.title("Sin(t)")
```

1.2 Generate and plot sinusoidal signals of 1V amplitude, 5Hz frequency with 0,90,180,270 phases in degrees.

```python
a=1
f1=5
phi1 = 0
phi2 = (np.pi)/2
phi3 = np.pi
phi4 = (3*(np.pi))/2

plt.plot(t,a*np.sin(2*np.pi*f1*t))
plt.xlabel("time(t)")
plt.ylabel("Amplitude")
plt.title("Sin(t) with 0 phase shift")
```

```python

plt.plot(t,a*np.sin(2*np.pi*f1*t + phi4))
plt.xlabel("time(t)")
plt.ylabel("Amplitude")
plt.title("Sin(t) with 270 phase shift")
```

```python

plt.plot(t,a*np.sin(2*np.pi*f1*t + phi3))
plt.xlabel("time(t)")
plt.ylabel("Amplitude")
plt.title("Sin(t) with 180 phase shift")
```

```python

plt.plot(t,a*np.sin(2*np.pi*f1*t + phi2))
plt.xlabel("time(t)")
plt.ylabel("Amplitude")
plt.title("Sin(t) with 90 phase shift")
```

1.3 Generate and plot a three phase sine wave with 1V amplitude and 5Hz frequency.

```python
y1 = np.sin(2*np.pi*5*t)
y2 = np.sin(2*np.pi*5*t-2 * np.pi / 3)
y3 = np.sin(2*np.pi*5*t-4 * np.pi / 3)

plt.plot(t,y1,'r')
plt.plot(t,y2,'b')
plt.plot(t,y3,'g')

plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("Three phase sine wave")
```

1.4 Generate and plot real exponential function signals of the form e^at with a=2 and -2.

```python
y4 = np.exp(2*t)
y5 = np.exp(-2*t)

plt.plot(t,y4)
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("Plot of e^at")
```

```python
plt.plot(t,y5)
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("Plot of e^(-at)")
```

1.5 Generate and plot the forward characteristics of PN junction diode.

```python
I_o = 12
V = np.linspace(-10,10,1000)
V_t = 0.025

y6 = I_o*(np.exp(V/V_t)-1)

plt.plot(V,y6)
plt.xlabel("Voltage(V)")
plt.ylabel("Current(I)")
plt.title("Forward characteristics of PN junction diode")
```

1.6 Generate and plot the function for radioactive decay.

```python
N0 = 1000        
half_life = 2.0  
lam = np.log(2) / half_life  

t2 = np.linspace(0, 10, 500)
N = N0 * np.exp(-lam * t2)

plt.plot(t2,N)
plt.xlabel("Time(t)")
plt.ylabel("Undecayed(N)")
plt.title(" Graph for function of radioactive decay.")
```

1.7 Generate and plot the complex exponential signals.

```python
v1 = np.exp(1j*2*np.pi*5*t)

plt.subplot(2,1,1)
plt.plot(t,np.real(v1))
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")


plt.subplot(2,1,2)
plt.plot(t,np.imag(v1))
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
```

1.8 Generate and plot the sine and cosine waves.

```python
plt.plot(t,np.sin(2*np.pi*5*t))
plt.plot(t,np.cos(2*np.pi*5*t))
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("Sine and Cosine waves")
```

1.9 Generate and plot the outcome of multiplying sinusoidal signals with an exponential signal.

```python
f1 = np.exp(2*t)
f2 = f1*np.sin(2*np.pi*5*t)
f3 = np.exp(-2*t)*np.sin(2*np.pi*5*t)

plt.plot(t,f2)
plt.plot(t,f3)
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("Multiplication of sinusoidal signals with an exponential signal")
```

1.10 Generate and compare the sine wave with variable t as t^2 and comment on the results.

```python
p1 = np.sin(2*np.pi*5*t)
p2 = np.sin(2*np.pi*5*(t**2))

plt.plot(t,p1,'r')
plt.plot(t,p2,'b')
plt.xlabel("Time(t)")
plt.ylabel("Amplitude")
plt.title("sine wave with variable t as t^2")
```
